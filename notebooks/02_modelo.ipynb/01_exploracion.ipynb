```python
# 01_exploracion.ipynb
# 
# Este notebook contiene el análisis exploratorio de los datos

# %% [markdown]
# # Análisis Exploratorio de Datos - ENEMDU 2025
# 
# ## 1. Carga de Datos

# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar los datos
df = pd.read_csv('data/processed/enemdu_2025_limpio.csv')
print(f"Registros: {len(df):,}")
print(f"Variables: {len(df.columns)}")

# %% [markdown]
# ## 2. Estadísticas Descriptivas

# %%
# Estadísticas básicas
df.describe()

# %%
# Distribución de la variable dependiente
print("Distribución de Informalidad:")
print(df['informal'].value_counts(normalize=True))

# %% [markdown]
# ## 3. Análisis Gráfico

# %%
# Tasa de informalidad por grupos
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Por género
df.groupby('sexo')['informal'].mean().plot(kind='bar', ax=axes[0,0])
axes[0,0].set_title('Informalidad por Género')

# Por edad
df.groupby(pd.cut(df['edad'], bins=range(15, 70, 5)))['informal'].mean().plot(ax=axes[0,1])
axes[0,1].set_title('Informalidad por Edad')

plt.tight_layout()
plt.savefig('outputs/figures/exploratorio.png', dpi=300, bbox_inches='tight')
plt.show()
```
